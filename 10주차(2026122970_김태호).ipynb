```python
result = 0

def adder(num):
    global result
    result += num
    return result

print("--- 계산기 1대 ---")
print(adder(3))
print(adder(4))

result1 = 0
result2 = 0

def adder1(num):
    global result1
    result1 += num
    return result1

def adder2(num):
    global result2
    result2 += num
    return result2

print("\n--- 계산기 2대 ---")
print("계산기1:", adder1(3))
print("계산기1:", adder1(4))
print("계산기2:", adder2(3))
print("계산기2:", adder2(7))
```

```python
class Calculator:
    def __init__(self):
        self.result = 0

    def adder(self, num):
        self.result += num
        return self.result


cal1 = Calculator()
cal2 = Calculator()

print("--- 클래스를 이용한 계산기 ---")
print("cal1 결과:", cal1.adder(3))
print("cal1 결과:", cal1.adder(4))
print("cal2 결과:", cal2.adder(3))
print("cal2 결과:", cal2.adder(7))
```

```python
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second

    def sum(self):
        result = self.first + self.second
        return result

    def mul(self):
        result = self.first * self.second
        return result

    def sub(self):
        result = self.first - self.second
        return result

    def div(self):
        result = self.first / self.second
        return result


a = FourCal()
b = FourCal()

a.setdata(4, 2)
b.setdata(3, 7)

print("--- FourCal 연산 결과 ---")
print("a 덧셈:", a.sum())
print("a 곱셈:", a.mul())
print("a 뺄셈:", a.sub())
print("a 나눗셈:", a.div())
print("b 뺄셈:", b.sub())
```

```python
from random import randint

print("--- 순차적 주사위 게임 ---")
player1_dice = []
player2_dice = []

for i in range(3):
    player1_dice.append(randint(1, 6))
    player2_dice.append(randint(1, 6))

print("Player 1 rolled", player1_dice)
print("Player 2 rolled", player2_dice)

if sum(player1_dice) == sum(player2_dice):
    print("Draw")
elif sum(player1_dice) > sum(player2_dice):
    print("Player 1 wins")
else:
    print("Player 2 wins")


print("\n--- OOP 주사위 게임 ---")
class Player:
    def __init__(self):
        self.dice = []

    def roll(self):
        self.dice = []
        for i in range(3):
            self.dice.append(randint(1, 6))

    def get_dice(self):
        return self.dice

player1 = Player()
player2 = Player()

player1.roll()
player2.roll()

print("Player 1 rolled", player1.get_dice())
print("Player 2 rolled", player2.get_dice())

if sum(player1.get_dice()) == sum(player2.get_dice()):
    print("Draw!")
elif sum(player1.get_dice()) > sum(player2.get_dice()):
    print("Player 1 wins!")
else:
    print("Player 2 wins!")
```

```python
class Customer(object):
    """ABC Bank의 고객 클래스"""

    def __init__(self, name):
        self.name = name

    def set_balance(self, balance=0.0):
        """초기 잔고 설정"""
        self.balance = balance

    def withdraw(self, amount):
        """출금: 출금 후 잔고 반환"""
        if amount > self.balance:
            raise RuntimeError('Amount greater than available balance.')
        self.balance -= amount
        return self.balance

    def deposit(self, amount):
        """입금: 입금 후 잔고 반환"""
        self.balance += amount
        return self.balance


customer1 = Customer("홍길동")
customer1.set_balance(10000)

print(f"{customer1.name}님의 초기 잔고:", customer1.balance)
print("5000원 입금 후 잔고:", customer1.deposit(5000))
print("3000원 출금 후 잔고:", customer1.withdraw(3000))
```
