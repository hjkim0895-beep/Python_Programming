```python
# 기존 for문 방식
number_list = []
for number in range(1, 6):
    number_list.append(number)
print("for문 사용:", number_list)

# List Comprehension 방식 (더 간결함)
number_list2 = [number for number in range(1, 6)]
print("List Comprehension:", number_list2)
```

```python
# 수식 적용
print("수식 적용:", [number-1 for number in range(1, 6)])

# 조건(if) 적용 (홀수만 추출)
print("조건 적용:", [number for number in range(1, 6) if number % 2 == 1])

# 문자열 리스트에 함수 및 조건 적용
sentence = ['I', 'Love', 'Python', 'Soooooo', 'MUCH!!!']
print("소문자 변환:", [word.lower() for word in sentence])
print("길이 6 초과 단어:", [word for word in sentence if len(word) > 6])
```

```python
rows = range(1, 4)
cols = range(1, 3)

# 이중 for문을 한 줄로 작성
cells = [(row, col) for row in rows for col in cols]
print("이중 for문 결과:", cells)

# 변수의 범위가 앞의 변수에 의존하는 경우
print("i와 j 조합:", [(i, j) for i in range(5) for j in range(i)])
```

```python
word = 'letters'

# Dictionary Comprehension (set을 이용해 중복 문자 카운트 효율성 개선)
letter_counts = {letter: word.count(letter) for letter in set(word)}
print("Dictionary Comprehension:", letter_counts)

# Set Comprehension (중복 없는 집합 생성)
a_set = {number for number in range(1, 6) if number % 3 == 1}
print("Set Comprehension:", a_set)

# Dictionary의 Key와 Value 위치 바꾸기
d = {'a': 1, 'b': 2, 'c': 3}
print("Key와 Value 뒤집기:", {v:k for k, v in d.items()})
```

```python
# 덧셈 함수 정의
def sum(a, b):
    return a + b

print("정수 덧셈:", sum(1, 2))
print("실수 덧셈:", sum(1.3, 3.1))
print("문자열 덧셈:", sum('love ', 'python'))
```

```python
days = ['Monday', 'Tuesday', 'Wednesday']
fruits = ['banana', 'orange', 'peach']

# 여러 리스트를 동시에 순회
for day, fruit in zip(days, fruits):
    print(day, "- eat", fruit)

# zip을 이용하여 딕셔너리 만들기
english = ['Monday', 'Tuesday', 'Wednesday']
french = ['Lundi', 'Mardi', 'Mercredi']
print("영어-프랑스 사전:", dict(zip(english, french)))
```

```python
def menu(wine, entree, dessert):
    return {'wine': wine, 'entree': entree, 'dessert': dessert}

# 위치 인자 (순서대로 들어감)
print("위치 인자:", menu('chardonnay', 'chicken', 'cake'))

# 키워드 인자 (순서가 달라도 변수명으로 찾아감)
print("키워드 인자:", menu(entree='beef', dessert='bagel', wine='bordeaux'))
```

```python
# dessert에 기본값 'pudding' 설정
def menu_default(wine, entree, dessert='pudding'):
    return {'wine': wine, 'entree': entree, 'dessert': dessert}

# dessert를 생략하면 기본값이 들어감
print("기본값 사용:", menu_default('chardonnay', 'chicken'))

# 값을 명시하면 덮어씀
print("기본값 덮어쓰기:", menu_default('dunkelfelder', 'duck', 'doughnut'))
```

```python
def echo(anything):
    'echo returns its input argument'
    return anything

# 함수의 설명서(docstring) 출력하기
print("Docstring 출력 1:")
help(echo)

print("\nDocstring 출력 2:")
print(echo.__doc__)
```
