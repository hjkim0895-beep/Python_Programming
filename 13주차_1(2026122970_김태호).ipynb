```python

xml_data = """<menu>
  <breakfast hours="7-11">
    <item price="$6.00">breakfast burritos</item>
    <item price="$4.00">pancakes</item>
  </breakfast>
  <lunch hours="11-3">
    <item price="$5.00">hamburger</item>
  </lunch>
</menu>"""

with open('menu.xml', 'w', encoding='utf-8') as f:
    f.write(xml_data)
print("menu.xml 파일 생성 완료!")
```

```python
import xml.etree.ElementTree as et

tree = et.ElementTree(file='menu.xml')
root = tree.getroot()

print("최상위 태그 이름:", root.tag)
print("첫 번째 자식(breakfast)의 태그와 속성:", root[0].tag, root[0].attrib)
print("menu의 자식 개수:", len(root))
```

```python

lunch = root.find("lunch")
print("lunch 태그의 hours 속성:", lunch.get("hours"))


print("\n[모든 item의 가격과 텍스트]")
for item in root[0].findall('item'): # breakfast 안의 item들
    print(f"가격: {item.get('price')}, 메뉴명: {item.text}")
```

```python
import json

info_data = {"name": "홍길동", "birth": "0525", "age": 30}
with open('./myinfo.json', 'w', encoding='utf-8') as f:
    json.dump(info_data, f, ensure_ascii=False)

with open('./myinfo.json', encoding='utf-8') as f:
    data = json.load(f)

print("파일에서 읽어온 데이터:", data)
```

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

html = urlopen("http://pythonscraping.com/pages/page1.html")
bsObj = BeautifulSoup(html.read(), "html.parser")

print("h1 태그 전체:", bsObj.h1)
print("h1 태그 내부의 텍스트만:", bsObj.h1.text)
print("\n계층적 접근 (bsObj.html.body.h1):", bsObj.html.body.h1)
```
